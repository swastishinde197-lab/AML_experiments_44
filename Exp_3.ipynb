```python
import pandas as pd
import numpy as np

from google.colab import files
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.feature_selection import SequentialFeatureSelector
```

```python
uploaded = files.upload()
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
df.head()
```

```python
le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])
```

```python
X = df[['Gender', 'Age', 'EstimatedSalary']]
y = df['Purchased']
```

```python
scaler = StandardScaler()
X = scaler.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

```python
model = LogisticRegression()
```

```python
sfs = SequentialFeatureSelector(
    model,
    n_features_to_select=2,
    direction='forward',
    scoring='accuracy',
    cv=5
)

sfs.fit(X_train, y_train)

sfs_features = sfs.get_support()
selected_sfs = np.array(['Gender', 'Age', 'EstimatedSalary'])[sfs_features]

print("Selected Features using SFS:", selected_sfs)
```

```python
X_train_sfs = sfs.transform(X_train)
X_test_sfs = sfs.transform(X_test)

model.fit(X_train_sfs, y_train)
y_pred_sfs = model.predict(X_test_sfs)

accuracy_sfs = accuracy_score(y_test, y_pred_sfs)
print("Accuracy using SFS:", accuracy_sfs)
```

```python
sbe = SequentialFeatureSelector(
    model,
    n_features_to_select=2,
    direction='backward',
    scoring='accuracy',
    cv=5
)

sbe.fit(X_train, y_train)

sbe_features = sbe.get_support()
selected_sbe = np.array(['Gender', 'Age', 'EstimatedSalary'])[sbe_features]

print("Selected Features using SBE:", selected_sbe)
```

```python
X_train_sbe = sbe.transform(X_train)
X_test_sbe = sbe.transform(X_test)

model.fit(X_train_sbe, y_train)
y_pred_sbe = model.predict(X_test_sbe)

accuracy_sbe = accuracy_score(y_test, y_pred_sbe)
print("Accuracy using SBE:", accuracy_sbe)
```

```python
print("SFS Accuracy:", accuracy_sfs)
print("SBE Accuracy:", accuracy_sbe)
```

```python
print("Selected Features using SFS:", selected_sfs)
print("Accuracy using SFS:", accuracy_sfs)

print("Selected Features using SBE:", selected_sbe)
print("Accuracy using SBE:", accuracy_sbe)
```
