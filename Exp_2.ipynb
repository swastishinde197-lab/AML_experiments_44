```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.feature_selection import VarianceThreshold, SelectKBest
from sklearn.feature_selection import chi2, f_classif, mutual_info_classif
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
df.head()
```

```python
le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])
```

```python
X = df[['Gender', 'Age', 'EstimatedSalary']]
y = df['Purchased']
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(X_scaled, columns=X.columns)
X_scaled.head()
```

```python
vt = VarianceThreshold(threshold=0.1)
vt.fit(X_scaled)

variance_features = X.columns[vt.get_support()]
print("Selected features (Variance Threshold):", list(variance_features))
```

```python
corr = df.corr()['Purchased'].abs().sort_values(ascending=False)
corr_features = corr.index[1:3]

print("Top features (Correlation):", list(corr_features))
```

```python
X_chi = X.copy()
X_chi['Age'] = abs(X_chi['Age'])
X_chi['EstimatedSalary'] = abs(X_chi['EstimatedSalary'])

chi_selector = SelectKBest(score_func=chi2, k=2)
chi_selector.fit(X_chi, y)

chi_features = X.columns[chi_selector.get_support()]
print("Top features (Chi-Square):", list(chi_features))
```

```python
anova_selector = SelectKBest(score_func=f_classif, k=2)
anova_selector.fit(X_scaled, y)

anova_features = X.columns[anova_selector.get_support()]
print("Top features (ANOVA):", list(anova_features))
```

```python
mi_scores = mutual_info_classif(X_scaled, y, random_state=42)

mi_df = pd.Series(mi_scores, index=X.columns).sort_values(ascending=False)
print("Top features (Mutual Information):")
print(mi_df.head(2))
```
